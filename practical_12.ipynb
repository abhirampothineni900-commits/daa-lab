```python

import time
from itertools import permutations

# Travelling Salesman Problem
def tsp(graph, n):

    start_time = time.perf_counter()

    cities = list(range(1, n))
    min_cost = float('inf')
    best_path = []

    for path in permutations(cities):
        current_cost = 0
        current_city = 0

        for city in path:
            current_cost += graph[current_city][city]
            current_city = city

        # Return to starting city
        current_cost += graph[current_city][0]

        if current_cost < min_cost:
            min_cost = current_cost
            best_path = [0] + list(path) + [0]

    end_time = time.perf_counter()

    execution_time = end_time - start_time

    return best_path, min_cost, execution_time


# User input
n = int(input("Enter number of cities: "))

print("Enter the cost adjacency matrix:")

graph = []

for i in range(n):
    row = list(map(int, input().split()))
    graph.append(row)


# Execute TSP
path, cost, execution_time = tsp(graph, n)


# Display result
print("\nOptimal Travelling Salesman Path:")

for city in path:
    print(chr(65 + city), end=" ")

print("\nMinimum Travel Cost:", cost)


# Time complexity
print("\nTime Complexity Analysis:")
print("Best Case: O(V!)")
print("Average Case: O(V!)")
print("Worst Case: O(V!)")
print("Space Complexity: O(V)")


# Execution time
print("\nActual Execution Time:")
print(f"{execution_time:.8f} seconds")
print(f"{execution_time * 1000:.6f} milliseconds")
```
