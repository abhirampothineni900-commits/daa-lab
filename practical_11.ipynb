```python

import time

# Floyd-Warshall Algorithm
def floyd_warshall(graph, n):
    dist = [row[:] for row in graph]

    start_time = time.perf_counter()

    for k in range(n):
        for i in range(n):
            for j in range(n):
                if dist[i][k] != float('inf') and dist[k][j] != float('inf'):
                    dist[i][j] = min(
                        dist[i][j],
                        dist[i][k] + dist[k][j]
                    )

    end_time = time.perf_counter()

    execution_time = end_time - start_time

    return dist, execution_time


# Taking user input
n = int(input("Enter number of vertices: "))

print("Enter adjacency matrix:")
print("Enter 99999 if there is no direct edge.")

graph = []

for i in range(n):
    row = list(map(int, input().split()))

    for j in range(n):
        if row[j] == 99999:
            row[j] = float('inf')

    graph.append(row)


# Execute algorithm
result, execution_time = floyd_warshall(graph, n)


# Display shortest distance matrix
print("\nShortest Distance Matrix:")

for row in result:
    for value in row:
        if value == float('inf'):
            print("INF", end="\t")
        else:
            print(value, end="\t")
    print()


# Display time complexity
print("\nTime Complexity Analysis:")
print("Best Case: O(V^3)")
print("Average Case: O(V^3)")
print("Worst Case: O(V^3)")
print("Space Complexity: O(V^2)")

# Display actual execution time
print("\nActual Execution Time:")
print(f"{execution_time:.8f} seconds")
print(f"{execution_time * 1000:.6f} milliseconds")
```
